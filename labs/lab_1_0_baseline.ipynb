### Task 1: Environment Sanity Check
Run the cell below to verify your Python version and library availability.

```python
import sys
import pandas as pd
import requests
print(f"Python Version: {sys.version}")
print(f"Pandas Version: {pd.__version__}")
print("✅ Substrate libraries loaded.")
```

### Task 2: API Connectivity (Ollama Check)
This cell pings the local Ollama server running inside your container. If this fails, your `postStartCommand` in `devcontainer.json` did not execute correctly.

```python
try:
    response = requests.get("http://localhost:11434/api/tags")
    if response.status_code == 200:
        print("✅ Ollama is alive and responding.")
        print("Available Models:", response.json().get('models', []))
    else:
        print("❌ Ollama responded with an error.")
except Exception as e:
    print(f"❌ Connection Failed: {e}")
```

### Task 3: The First Inference
Execute a basic generation request. This will pull the 'mistral' or 'llama3' model if it isn't already present.

```python
payload = {
    "model": "mistral",
    "prompt": "The first rule of LLM engineering is:",
    "stream": False
}
response = requests.post("http://localhost:11434/api/generate", json=payload)
print("LLM Response:", response.json()['response'])
```
